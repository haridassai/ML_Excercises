```python
import numpy as np
import pandas as pd

data = pd.DataFrame({
    "age": [25, 30, np.nan, 45, 50],
    "income": [40000, 50000, 60000, np.nan, 100000],
    "city": ["Hyderbad", "Chennai", None, "Hyderabad", None]
})

```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer

numeric_column = ["age", "income"]
categorical_column = ["city"]

preprocessor = ColumnTransformer(
    transformers =[
        (
            "numeric_imputer",
            SimpleImputer(strategy="median"),
            numeric_column
        ),
        (
            "categorical_imputer",
            SimpleImputer(
                strategy="constant",
                fill_value="Unknown"
            ),
            categorical_column
        )
    ],
    verbose_feature_names_out=False
)

transformed_array = preprocessor.fit_transform(data)

transformed_data = pd.DataFrame(
    transformed_array,
    columns=preprocessor.get_feature_names_out(),
    index=data.index
)

print(transformed_data)
```
